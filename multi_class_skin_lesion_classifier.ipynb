```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dropout, BatchNormalization, Flatten, Dense
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os
import numpy as np
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.models import load_model
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Ensure necessary directories exist
os.makedirs("saved_models", exist_ok=True)

# No validation_split needed here
train_datagen = ImageDataGenerator(rescale=1./255)
val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    '/content/unzipped_folder/skin_data_split/train',
    target_size=(96, 96),
    batch_size=32,
    class_mode='categorical'
)

val_generator = val_datagen.flow_from_directory(
    '/content/unzipped_folder/skin_data_split/val',
    target_size=(96, 96),
    batch_size=32,
    class_mode='categorical'
)
test_generator = test_datagen.flow_from_directory(
    '/content/unzipped_folder/skin_data_split/test',
    target_size=(96, 96),
    batch_size=1,  # important to set 1 for prediction
    class_mode='categorical',
    shuffle=False
)

# Get the number of classes found by the generator
num_classes = train_generator.num_classes
print(f"Number of classes found by generator: {num_classes}")

model = Sequential()

# Block 1
model.add(Conv2D(32, (3,3), activation='relu', padding='same', input_shape=(96, 96, 3)))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(3,3)))
model.add(Dropout(0.25))

# Block 2
model.add(Conv2D(64, (3,3), activation='relu', padding='same'))
model.add(Conv2D(64, (3,3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2)))
model.add(Dropout(0.25))

# Block 3
model.add(Conv2D(128, (3,3), activation='relu', padding='same'))
model.add(Conv2D(128, (3,3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2)))
model.add(Dropout(0.25))

# Fully connected
model.add(Flatten())
model.add(Dense(1024, activation='relu'))
model.add(BatchNormalization())
model.add(Dropout(0.5))
# Update the dense layer to match the number of classes found by the generator
model.add(Dense(num_classes, activation='softmax'))

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss='categorical_crossentropy',
              metrics=['accuracy'])


# Get class names from generator
class_names = list(train_generator.class_indices.keys())

# Count images in each class folder
class_counts = [len(os.listdir(os.path.join('/content/unzipped_folder/skin_data_split/train', cls))) for cls in class_names]

# Create label array (needed for class_weight computation)
# Ensure labels array covers all unique class indices found by the generator
labels = np.concatenate([[i]*count for i, count in enumerate(class_counts)])

# Compute class weights
class_weights_array = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(labels),
    y=labels
)

class_weights = dict(enumerate(class_weights_array))
print("✅ Class Weights Used:", class_weights)


# Define callbacks
early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

checkpoint = ModelCheckpoint(
    filepath='saved_models/best_cnn_model.h5',
    monitor='val_accuracy',
    save_best_only=True,
    verbose=1
)
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=150,
    class_weight=class_weights,
    callbacks=[early_stop, checkpoint]
)

# Load the best model (optional, but good practice)
# model = load_model('saved_models/best_cnn_model.h5') # This line is not needed here if you continue with the trained model


# Predict
Y_pred = model.predict(test_generator, verbose=1)
y_pred = np.argmax(Y_pred, axis=1)

# True labels
y_true = test_generator.classes
class_labels = list(test_generator.class_indices.keys()) # Get labels from test generator for consistency

# Classification Report
print("Classification Report:")
print(classification_report(y_true, y_pred, target_names=class_labels))

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=class_labels, yticklabels=class_labels)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

# Accuracy and loss plot
plt.plot(history.history['accuracy'], label='train acc')
plt.plot(history.history['val_accuracy'], label='val acc')
plt.legend()
plt.title('Model Accuracy')
plt.show()

plt.plot(history.history['loss'], label='train loss')
plt.plot(history.history['val_loss'], label='val loss')
plt.legend()
plt.title('Model Loss')
plt.show()
```

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# The model is already loaded in the previous cell from '/content/drive/MyDrive/saved_models/best_cnn_model.h5'

# Upload an image
uploaded = files.upload()

# Define class labels (update this based on your training classes if needed)
# It's safer to get the class labels from the test_generator if you can access it here,
# but assuming the labels are consistent:
# class_labels = list(test_generator.class_indices.keys())
# If test_generator is not available here, use the hardcoded list, but ensure it matches your training classes.
class_labels = ['acne', 'akiec', 'bags', 'bcc', 'bkl', 'df', 'mel', 'nv', 'redness', 'vasc'] # **Verify these labels match your actual classes**

# Loop over uploaded files
for file_name in uploaded.keys():
    # Load and preprocess the image
    img = load_img(file_name, target_size=(96, 96))  # match input shape
    img_array = img_to_array(img)
    img_array = img_array / 255.0  # rescale as during training
    img_array = np.expand_dims(img_array, axis=0)  # add batch dimension

    # Predict using the already loaded model
    predictions = model.predict(img_array)
    predicted_class_index = np.argmax(predictions)

    # Ensure the predicted_class_index is within the bounds of class_labels
    if 0 <= predicted_class_index < len(class_labels):
        predicted_class = class_labels[predicted_class_index]
    else:
        predicted_class = "Unknown Class" # Handle cases where prediction index is out of range
        print(f"Warning: Predicted class index {predicted_class_index} out of bounds for class_labels.")


    # Display results
    print(f"🔍 Predicted class for {file_name}: {predicted_class}")
    plt.imshow(img)
    plt.title(f"Predicted: {predicted_class}")
    plt.axis('off')
    plt.show()
```
